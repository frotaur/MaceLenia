```python
from modules import MaCELenia, FlowLenia
from modules.main_utils import print_screen
h=300
w=300
device='cuda'
from pathlib import Path
cur_dir = Path('.')

mace=  MaCELenia(
            (1, h, w),
            dt=0.1,
            num_channels=3,
            device=device,
            has_food=False,
            # save_dir="saved_macelenia",
            save_dir="compare_mace",
            interest_files=(cur_dir / "compare_mace").as_posix(),
        )

flow = FlowLenia(
            (1, h, w),
            dt=0.2,
            dd=5,
            num_channels=3,
            save_dir="saved_flow_lenia",
            interest_files=(cur_dir / "compare_mace").as_posix(),
            device=device,
            has_food=False,
        )
```

```python
num_steps = 1000
from tqdm import tqdm
for k in tqdm(range(len(mace.interest_files))):
    mace.load_interest_file()
    flow.load_interest_file()
    for i in range(num_steps):
        mace.step()
        flow.step()
    mace.draw()
    flow.draw()
    print_screen(mace.worldsurface, f'mace_{k}')
    print_screen(flow.worldsurface, f'flow_{k}')
```
